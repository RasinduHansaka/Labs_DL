```python


# 1. Install required libraries (only needed in Colab)
!pip install seaborn pandas matplotlib numpy --quiet

# 2. Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 3. Load your dataset
file_path = "/content/Au_nanoparticle_dataset.csv"  # Upload file to Colab
df = pd.read_csv(file_path)

# Display first few rows
print("First 5 rows of the dataset:")
print(df.head())

# 4. Pearson & Spearman Correlations
# Selecting numeric columns only
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

pearson_corr = df[numeric_cols].corr(method="pearson")
spearman_corr = df[numeric_cols].corr(method="spearman")

# 5. Plot Pearson correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(pearson_corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Pearson Correlation Heatmap", fontsize=16)
plt.show()

# 6. Plot Spearman correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(spearman_corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Spearman Correlation Heatmap", fontsize=16)
plt.show()
```

```python


# 1. Install required libraries (only needed in Colab)
!pip install seaborn pandas matplotlib numpy --quiet

# 2. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 3. Load dataset
file_path = "/content/Au_nanoparticle_dataset.csv"  # Upload file to Colab
df = pd.read_csv(file_path)

# Show dataset info
print("\nDataset Info:")
print(df.info())

# Show first few rows
print("\nFirst 5 rows:")
print(df.head())

# 4. Example plots (adjust column names based on your dataset)
# NOTE: Replace 'Column1', 'Column2', etc. with actual column names from df.columns

# Histogram for a numeric column
plt.figure(figsize=(8, 6))
sns.histplot(df[df.columns[0]], bins=20, kde=True)  # Example: first numeric column
plt.title(f"Histogram of {df.columns[0]}")
plt.show()

# Scatter plot between two numeric columns
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df.columns[0], y=df.columns[1], data=df)
plt.title(f"Scatter plot: {df.columns[0]} vs {df.columns[1]}")
plt.show()

# Box plot for a numeric column grouped by a categorical column (if available)
# Check if there's a categorical column
categorical_cols = df.select_dtypes(exclude=[float, int]).columns.tolist()
if categorical_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=categorical_cols[0], y=df.columns[0], data=df)
    plt.title(f"Boxplot of {df.columns[0]} by {categorical_cols[0]}")
    plt.show()
else:
    print("\nNo categorical columns found for boxplot.")
```
